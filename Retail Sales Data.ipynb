```python
train[["MarkDown1" ,"MarkDown2", "MarkDown3", "MarkDown4"]] = train[["MarkDown1" ,"MarkDown2", "MarkDown3", "MarkDown4"]].fillna(0)
```

```python
print(train[["MarkDown1" ,"MarkDown2", "MarkDown3", "MarkDown4"]])
```

Finding Outliers

```python
Q1 = train['Weekly_Sales'].quantile(0.25)
Q3 = train['Weekly_Sales'].quantile(0.75)

IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

#Detect Outliers
num_lower_outliers = (train['Weekly_Sales'] < lower).sum()
lower_outliers = train[(train['Weekly_Sales'] < lower)]
num_upper_outliers = (train['Weekly_Sales'] > upper).sum()
upper_outliers = train[(train['Weekly_Sales'] > upper)]
print("Number of Lower Outliers are :- ",num_lower_outliers)
print(lower_outliers)
print("Number of Upper Outliers are :- ", num_upper_outliers)
print(upper_outliers)
```
